```python
# TASK

purchase_amount = int(input("Provide total cost for discount: "))
isLoyalityMember = True
dayOfWeek = input("Enter a day: ")
isHolidaySeason = True

print(f"Total Amount {purchase_amount}$")

if(purchase_amount > 100):
  purchaseDiscount =  purchase_amount  * 0.10
  print(f"Discount on purschasing {purchaseDiscount}$")
elif(purchase_amount > 50 or purchase_amount == 100 ):
  purchaseDiscount =  purchase_amount * 0.05
  print(f"Discount on purchasing {purchaseDiscount}$")
else:
  print("No purchasing discount")

if(isLoyalityMember == True):
    loyalityDiscount =  purchase_amount * 0.05
    print(f"Discount for loyality member {loyalityDiscount}$")
else:
  print("No loyality member discount")

if(dayOfWeek == "sunday" or dayOfWeek == "saturday"):
      weekdayDiscount =  purchase_amount * 0.02
      print(f"Discount on weekend {weekdayDiscount}$")
else:
  print("No weekend discount")

if(isHolidaySeason == True):
      holidayDiscount =  purchase_amount * 0.08
      print(f"Discount on holiday season {holidayDiscount}$")

else:
  print("No holiday season discount")

totalDiscount = purchaseDiscount + loyalityDiscount + weekdayDiscount + holidayDiscount
totalAmount = purchase_amount - totalDiscount
print(f"Total Discount : {totalDiscount}$ ")
print(f"Total Amount : {totalAmount}$ ")
```

```python
# WHILE LOOP

# GUESSING GAME

print("This is number guessing game ")
secretNumber = 10
guesses = 5
while guesses > 0:
  print(f"\n You have {guesses} guesses left.")

  guessingNUmber = int(input("Guess the number : "))
  if(guessingNUmber > secretNumber):
   print("Guessing number is too high")
  elif(guessingNUmber < secretNumber):
    print("Guessing number is too low")
  elif(secretNumber == guessingNUmber):
    print(f"Congratulation, you guessed the correct number {secretNumber}" )
    break
  else:
    print("Choose number only")
  guesses -= 1
  if guesses == 0:
      print("You are out of game")
```

```python
# STUDENT GRADING SYSTEM

students = []

students.append({
    "name" : "Bashar",
    "id" : 1,
    "grades" : {
        "Biology" : 79,
        "Physics" : 87,
        "Chemistry" : 80
    }
})
students.append({
    "name" : "Saad",
    "id" : 2,
    "grades" : {
        "Biology" : 65,
        "Physics" : 87,
        "Chemistry" : 79
    }
})
students.append({
    "name" : "Zohaib",
    "id" : 3,
    "grades" : {
        "Biology" : 87,
        "Physics" : 69,
        "Chemistry" : 79
    }
})

print("Students Information")

for std in students:
  print(f"Name: {std['name']}")
  print(f"ID: {std['id']}")
  print("Grades: " )
  for subject, grade in std["grades"].items():
    print(f"{subject}: {grade}")
  print("-" * 20)
```
